```python
# Cantidad de nulos por columna
nulos = df.isnull().sum()

# Porcentaje de nulos por columna
porcentaje_nulos = (df.isnull().mean() * 100).round(2)

# Juntamos ambos en una tabla, ordenada de mayor a menor
tabla_nulos = pd.DataFrame({
    "cantidad_nulos": nulos,
    "porcentaje_nulos": porcentaje_nulos
}).sort_values("porcentaje_nulos", ascending=False)

# Nos quedamos solo con columnas que tienen al menos un nulo
tabla_nulos = tabla_nulos[tabla_nulos["cantidad_nulos"] > 0]

print(f"Columnas con nulos: {len(tabla_nulos)} de {df.shape[1]}")
tabla_nulos.head(20)
```

```python
diccionario = pd.read_csv(
    "../datos/crudos/HomeCredit_columns_description.csv",
    encoding="latin-1"
)

print(diccionario.shape)
diccionario.head()
```

```python
dic_app = diccionario[diccionario["Table"] == "application_{train|test}.csv"].copy()

# Limpiamos espacios invisibles en los nombres
dic_app["Row"] = dic_app["Row"].str.strip()

print(len(dic_app))
```

```python
pd.set_option("display.max_colwidth", None)  # que no corte las descripciones largas

def describir(columna):
    fila = dic_app[dic_app["Row"] == columna]
    return fila[["Row", "Description", "Special"]]

describir("TARGET")
```

```python
for col in ["DAYS_BIRTH", "DAYS_EMPLOYED", "AMT_CREDIT", "AMT_ANNUITY", "AMT_GOODS_PRICE","AMT_INCOME_TOTAL", "AMT_CREDIT", "AMT_ANNUITY"]:
    display(describir(col))
```
